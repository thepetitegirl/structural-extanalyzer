# Extraction: model selection and prompt engineering

Two considerations:

1. **Which model can run to extract the required data?** Langchain uses `with_structured_output` forces a tool call, which not every model supports. Furthermore, choice of model and model provider used may also impact the quality of the output and the cost 
2. **Does the prompt get the right values?** Design for prompt engineering may affect the type of output generated

Ground truth comes from `01_value_exploration.ipynb`; scoring uses
`evaluation/expected.yaml` via `src/evaluation.py`. Assumptions are at the end.

# Part 1: Which model can run this

## The constraint that comes first

**Free to run, no card, quick to start.** A project constraint, not a finding,
stated first because it eliminates candidates before any testing.

GPT-4 and Claude have the most reliable tool-calling available and would very
likely work first try. They are excluded because they are paid, not because they
failed. Everything below is a search within the free tier.

| Stage | Eliminates | Basis |
|---|---|---|
| 1. Free, no card | OpenAI, Anthropic | project constraint, not tested |
| 2. Tool-calling works | gpt-oss 20B and 120B | measured - tool_use_failed |
| 3. Returns correct values | local Ollama models | measured - returned 28400000000.0 |
| 4. Choice among survivors | Gemini free tier | preference for open weights |

Three axes: **open vs closed source**, **provider** (local weights vs hosted
key), and **LangChain compatibility**. The third decides, and it eliminates the
candidate that looks strongest on the other two.

````python
import os
import subprocess
import sys
import time
from pathlib import Path

from dotenv import load_dotenv
from pydantic import BaseModel, Field

sys.path.insert(0, str(Path.cwd().parent))
load_dotenv(Path.cwd().parent / ".env")

from src.llm import get_chat_model


class SingleValue(BaseModel):
    """Minimal schema. If forced tool-calling works at all, it works here."""

    value: float = Field(description="the number")


PROBE = "The tax was $28.4 billion. What is the value?"

for key in ("GROQ_API_KEY", "GOOGLE_API_KEY"):
    print(f"{key}: {'set' if os.environ.get(key) else 'not set - those cells will skip'}")
````

## The specification: structured output

`with_structured_output` does not ask the model to write JSON. It converts the
Pydantic schema into a tool signature and forces the model to call it, so the
response is validated rather than parsed hopefully.

That is a stronger specification than chat, and not every model meets it. The
probe below uses a one-field schema - about as easy as forced tool-calling gets.
A model that fails this cannot run the extraction at all.

````python
def probe(provider, model, **kwargs):
    """Run the one-field probe and report value, latency, or the failure."""
    started = time.perf_counter()
    try:
        answer = (
            get_chat_model(provider, model, **kwargs)
            .with_structured_output(SingleValue)
            .invoke(PROBE)
        )
        elapsed = time.perf_counter() - started
        # 28.4 is correct; 28400000000.0 means the model expanded the unit
        verdict = "Pass" if abs(answer.value - 28.4) < 0.01 else "Wrong"
        return verdict, answer.value, elapsed, ""
    except Exception as exc:
        return "Fail", None, time.perf_counter() - started, str(exc).split("{")[0][:60]
````

## Axis 1: open vs closed source

### Hosted open-source models, via Groq

````python
GROQ_MODELS = [
    "llama-3.3-70b-versatile",
    "llama-3.1-8b-instant",
    "openai/gpt-oss-20b",
    "openai/gpt-oss-120b",
]

groq_results = {}
if os.environ.get("GROQ_API_KEY"):
    for name in GROQ_MODELS:
        verdict, value, elapsed, error = probe("groq", name, temperature=0)
        groq_results[name] = (verdict, value, elapsed)
        detail = f"{value}" if value is not None else error
        print(f"  {verdict:5s} {name:26s} {elapsed:5.1f}s  {detail}")
else:
    print("skipped - GROQ_API_KEY not set")
````

**Both gpt-oss models fail**, at 20b and at 120b, with
`tool_use_failed: model did not call a tool` and an empty generation. Since size
does not change the outcome, this is a tool-calling limitation rather than a
capability gap - the models are otherwise strong.

This is the clearest illustration of why axis 3 decides: gpt-oss-120b is the
most capable open model tested here and cannot be used at all.

### Closed-source, via Google

Gemini's free tier needs a key but no card. Runs only if `GOOGLE_API_KEY` is set.

````python
GOOGLE_MODELS = ["gemini-2.0-flash", "gemini-2.5-flash"]

google_results = {}
if os.environ.get("GOOGLE_API_KEY"):
    for name in GOOGLE_MODELS:
        verdict, value, elapsed, error = probe("google", name, temperature=0)
        google_results[name] = (verdict, value, elapsed)
        detail = f"{value}" if value is not None else error
        print(f"  {verdict:5s} {name:26s} {elapsed:5.1f}s  {detail}")
else:
    print("skipped - GOOGLE_API_KEY not set")
````

## Axis 2: provider

### Local, via Ollama

Local models need no key and no network, and the document never leaves the
machine. The cost is disk and RAM: Ollama downloads the full weights and loads
them to answer.

The cell below reports what is installed and what each model occupies while
running.

````python
def ollama(*args):
    """Run an ollama subcommand, returning empty output if it is unavailable."""
    try:
        return subprocess.run(
            ["ollama", *args], capture_output=True, text=True, timeout=30
        ).stdout
    except (FileNotFoundError, subprocess.TimeoutExpired):
        return ""


listing = ollama("list")
print(listing or "ollama not available")

total_ram = int(subprocess.run(
    ["sysctl", "-n", "hw.memsize"], capture_output=True, text=True
).stdout.strip()) // 1024**3
print(f"machine RAM: {total_ram} GB")
````

````python
LOCAL_MODELS = ["qwen3:8b", "llama3.2:latest"]

local_results = {}
if listing:
    for name in LOCAL_MODELS:
        verdict, value, elapsed, error = probe("ollama", name, temperature=0)
        loaded = [
            line for line in ollama("ps").splitlines()[1:] if name.split(":")[0] in line
        ]
        ram = " ".join(loaded[0].split()[2:4]) if loaded else "?"
        local_results[name] = (verdict, value, elapsed, ram)
        detail = f"{value}" if value is not None else error
        print(f"  {verdict:5s} {name:20s} {elapsed:5.1f}s  RAM {ram:8s}  {detail}")
else:
    print("skipped - ollama not available")
````

Both local models call the tool successfully - but both return
**28400000000.0** rather than 28.4. They read "$28.4 billion" and expanded the
unit into the value.

That is not a crash, and it would not raise an exception in production. It is a
silently wrong number off by a factor of a billion. For this schema, where
`unit` is a separate field, it is a failure - and precisely the kind that
provenance and known-answer scoring exist to catch.

`qwen3:8b` is also slow (it emits reasoning tokens before answering), which is
an argument against it independent of correctness.

### What local costs

| | Local (Ollama / HuggingFace) | Hosted (Groq / Google) |
|---|---|---|
| Disk | full weights - 1.3-14 GB per model | none |
| RAM while running | 2.5-13 GB | none |
| Key required | no | yes (both free tiers) |
| Network | none | required |
| Rate limits | none | yes, but irrelevant at one request per run |
| Reproducible from a clone | needs a manual `ollama pull`, unpinned | one line in `.env` |
| Privacy | document stays local | document sent to the provider |

On a 16 GB machine, `gpt-oss-20b` would need roughly 13 GB of RAM and a 14 GB
download - workable but tight, and it would swap. Hosting it on Groq removes
both costs, which is why "open source" and "runs locally" are separate decisions
rather than one.

Smaller quantised builds (bfloat16, NF4, or small HuggingFace models) reduce the
RAM cost but not the reproducibility problem: a cloner still needs the weights,
and quantisation tends to weaken exactly the instruction-following that the
page-binding constraints depend on.

Reproducibility is the deciding factor here, and it favours hosted: a credential
in `.env` clones cleanly, while several GB of local state does not.

## Axis 3: LangChain compatibility

Every provider tested exposes the same `BaseChatModel` interface, so LangChain
itself never discriminates - `get_chat_model()` swaps them behind one function.

What differs is how `with_structured_output` is honoured underneath:

| Mechanism | Reliability | Models here |
|---|---|---|
| Native tool-calling | strongest | Llama 3.x on Groq, Gemini |
| JSON mode | weaker - valid JSON, no tool semantics | some local models |
| Prompt-and-hope | weakest | models with neither |

The failure mode matters as much as the rate. gpt-oss fails loudly with a 400,
which is easy to diagnose. The local models fail *silently* with a plausible
number - far more dangerous in a pipeline whose output is a float.

## Decision

**llama-3.3-70b-versatile on Groq.**

### How the search actually went

**1. Started with gpt-oss.** Open weights, strong benchmarks, and hosted on Groq
so it costs no local disk or RAM. On paper the best fit for an open-source
specification.

**2. It failed on tool-calling** - tool_use_failed, empty generation - at 20B
**and** at 120B. Scaling 6x changed nothing, which is the informative part: this
is not a capability limit that a bigger model fixes. with_structured_output
requires a model trained to emit tool calls, and that is a training decision
rather than a scale effect. A model can be excellent and still be unusable here.

**3. Llama worked, and for a reason.** Llama 3.x supports native tool-calling, so
both sizes passed the probe. Not luck, and not a close call - 8B and 70B both
returned 28.4 cleanly.

**4. Chose 70B over 8B** as a hedge. The extraction prompt asks the model to hold
several competing constraints at once - page binding, prose preference, column
headers, parenthesised negatives - and larger models generally follow
multi-constraint instructions more reliably. This one is judgement: the 8B model
was not scored against the full five-field extraction.

### Against the three axes

| Axis | Why it points here |
|---|---|
| Open vs closed | Open weights, so the model can be inspected or self-hosted later. No subscription; Groq's free tier covers one request per run. |
| Provider | Hosted, so no disk or RAM cost on a 16 GB machine, and a clone needs one line in .env rather than a multi-GB download. Rate limits are irrelevant at this volume, and latency is not a deliverable. |
| LangChain | Native tool-calling, so with_structured_output is reliable - the axis that eliminated everything else. |

### Rejected

- **gpt-oss-20b / 120b** - cannot call tools through Groq, so the pipeline cannot
  run at all. Failing identically at both sizes is what makes this a tool-calling
  finding rather than a capability one.
- **qwen3:8b, llama3.2 (local)** - both returned 28400000000.0 instead of 28.4,
  **silently**. No exception, just a number a billion times too large. Add
  2.5-5.6 GB RAM and an unpinned manual download.
- **llama-3.1-8b-instant** - works, and faster. Rejected on the multi-constraint
  hedge above, not on measurement.
- **Gemini free tier** - a working closed-source option. Rejected in favour of
  open weights, which was a stated preference rather than a technical result.
- **HuggingFace local** - same weight-download problem as Ollama with more setup.
  The Inference API adds cold starts without improving on Groq.

### Assumptions

- **Free-tier only.** Paid models were never tested. If cost were not a
  constraint, GPT-4 or Claude would be the straightforward choice.
- Judged on a one-field probe plus the real extraction, not a broad benchmark.
  Failing the probe proves unusability; passing it does not guarantee the full
  five-field schema succeeds.
- Latency is not a deliverable. One request per run, so 3s and 50s are
  equivalent - qwen3 is rejected on correctness, not speed.
- Model size did not predict success: the largest model tested (120B) failed
  while 8B models passed.
- The prompt was tuned against the chosen model.
- Provider behaviour changes. gpt-oss tool-calling on Groq may improve, and this
  notebook re-runs to check.

## Why pages are required, not merely convenient

"Corporate Income Tax" appears eight times across seven pages, carrying a
different value nearly every time. The cell below finds them all.

````python


reader = pypdf.PdfReader(str(pdf_path))
for page_no, page in enumerate(reader.pages, start=1):
    text = page.extract_text() or ""
    for match in re.finditer(r"Corporate Income Tax", text):
        excerpt = " ".join(text[match.start():match.start() + 95].split())
        print(f"  p{page_no:2d}: {excerpt}")
````

Feeding the whole document would leave the model choosing among:

| Page | Value | What it actually is |
|---|---|---|
| 5 | **28.4** | Revised FY2023, prose - **the answer** |
| 8 | 28.38 | the same figure, table precision |
| 8 | 23.07 / 24.26 | FY2022 actual / FY2023 estimated |
| 9 | 27.2% | share of Operating Revenue |
| 16 | 28.03 | Estimated **FY2024** - a different year |
| 26 | 28,380 / 28,029 | the same figures in $million |
| 27 | 3.9% | share of GDP |

Every one is a plausible answer to "the amount of Corporate Income Tax". They
differ by year, by unit, and by whether they are an amount or a proportion -
and nothing in the number itself says which. Only the surrounding text
disambiguates, which is why the page citation does the work: it eliminates
seven wrong answers before the model reads anything.

This is also why a bare float is not a usable output. `28.38` and `28.03` are
both defensible-looking; only the page and quote reveal which was read.

## Why an LLM rather than a regex

The five fields sit in four different shapes:

| Field | Shape in the document |
|---|---|
| Corporate Income Tax | prose - `revised to $28.4 billion` |
| Year-on-year change | prose - `$4.1 billion (17.0%) higher` |
| Total top-ups | table total - `Total 20,352` |
| Tax list | narrative spread over two pages |
| Fiscal position | table row - `OVERALL FISCAL POSITION 1.72 (0.35) (3.57)` |

A regex handles one shape; five shapes means five patterns to write and
maintain. The model reads all four with one prompt.

The risk this buys is **mis-selection**, not invention. Every value is present in
the supplied text - the model is copying, not recalling - but it can copy the
wrong one. That is exactly what happened below.

````python
# The line the model must read for the fiscal position. Five numbers, and only
# the header line elsewhere on the page says which is which.
for line in page_text.splitlines():
    if "OVERALL FISCAL POSITION" in line or "Corporate Income Tax 2" in line:
        print(line.strip())
````

### What changed in v2

Three edits. Each one fixes a specific failure above.

| # | Change | v1 said | v2 says | Fixes |
|---|---|---|---|---|
| 1 | Page citations declared **binding**, with the trap named | "from page 20" | "PAGE 20 ONLY... a value under a different marker is the WRONG value. Check the marker above your quote before answering." Plus: do not take the similarly-named row from another page. | top-ups |
| 2 | Prose preference **scoped** to the cited page | "prefer the prose" | "prefer the prose ON THE CITED PAGE... this preference never justifies leaving the cited page" | fiscal position |
| 3 | Instruction to read **to the end** | "list the taxes named" | "PAGES 5 AND 6 ONLY - this is the strictest constraint in this task... if a tax name appears only in a table on another page, it does not belong in this list." No expected count is stated: naming one would leak the answer and defeat the scorer's count check. | tax list |

The pattern: v1 stated the right pages but never said that a plausible value
elsewhere was *wrong*. The model treated the citations as preferences and
resolved them against other instructions - which is exactly what rule 2 did to
the fiscal position, since "prefer prose" and "read page 8" pointed in opposite
directions and nothing said which won.

**Wording is load-bearing.** While templating the page numbers, "Read both pages
to the end" became "Read those pages to the end" - and the tax list dropped from
13 back to 7, reproducibly at temperature 0. Restoring the emphatic phrasing
restored the 13. A one-word change moved a field from pass to fail.

## Conclusion

**All five fields extract correctly, each from the page cited for it.**

### What the failures taught

The interesting result is v1, not v2. All three of its failures had one cause:
**an instruction the model treated as a preference rather than a constraint.**
v1 named the right page for every field and still read the wrong one twice,
because nothing told it that a plausible value on another page was *wrong*
rather than merely second-best.

The fix was not more context or a better model. It was making the constraint
explicit, naming the specific trap the model had fallen into, and scoping the
competing rule ("prefer prose") so it could not override the constraint.

### What made the failures visible

Provenance. Every field carries the text it was read from, so v1's errors were
diagnosable in seconds - the quotes named pages 8 and 5 where 20 and 8 were
required. Without quotes, `24.32` and `-3.6` are plausible numbers that would
have needed manual checking against the source to catch.

This is why the schema carries page and quote per field rather than bare floats.

### Limits of this result

- **Known answers were available.** Scoring works here because the correct
  figures were established by reading the document first. A pipeline over
  unseen documents has no such check, and provenance becomes the only guard.
- **Restricted input does much of the work.** Feeding four cited pages rather
  than 37 removes most of the ambiguity before the model sees it. The prompt has
  not been tested against the full document.
- **Model-specific.** The prompt was tuned against `llama-3.3-70b-versatile`. A
  weaker model may need the constraints stated more heavily still.
- **One document.** Page citations, unit conventions, and the two-fiscal-year
  structure are all specific to this publication.

## Assumptions

- **Only the cited pages are read** (5, 6, 8, 20). The same term recurs across
  the document with different values - Corporate Income Tax appears on seven
  pages - so narrowing the input removes ambiguity the model would otherwise
  have to resolve. The specifications supply the page citations, so retrieval is
  already solved.
- **Each field is read from the page cited for it,** not from wherever the
  figure is most precise. Page 5 says "$28.4 billion"; page 8's table says
  28.38. The citation decides, so 28.4 is correct here.
- **Target year is Revised FY2023.** The fields are labelled 2024 but cite pages
  5 and 8, which report FY2023. The page citations were taken as authoritative.
  True FY2024 figures are on pages 13-16 and differ - Corporate Income Tax is
  28.03 there, with a year-on-year change of -1.2% rather than +17.0%.
- **The value must match the cited page exactly, in the form written there.**
  Page 5 says "$28.4 billion", so the answer is 28.4 with unit `billion` - not
  28.38 (page 8's table precision), not 28,400,000,000 (the unit expanded into
  the number), and not 28,400 (converted to million). The figure and its unit
  are separate fields precisely so the two cannot be silently combined. Local
  models failed exactly here, returning 28400000000.0.
- **Units are recorded, not converted.** Page 20 states $million while the other
  pages state $billion. Each value carries its unit; normalising would hide a
  1000x error.
- **Page bindings live in `config.yml`,** not in the prompt text, so changing a
  page changes it in one place. A field bound to a page that is never extracted
  fails at config load rather than silently returning a wrong answer.
- **Temperature is 0** so the same pages yield the same figures.
- **The prompt is document-specific.** Page citations, the two-fiscal-year
  structure, and the top-ups trap are all particular to this publication.
- **The prompt was tuned against `llama-3.3-70b-versatile`.** A weaker model may
  need the constraints stated more heavily still.
